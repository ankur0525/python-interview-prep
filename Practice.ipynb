````python
# 7. Writing FIBONACCI Series
fib = [0,1]
n = 5
for i in range(2, n):
    next_fib= fib[i-1] + fib[i-2]
    fib.append(next_fib)
print(f'The first {n} numbers in the Fibonacci series are: {fib}')
````

## Why This Code is Correct - Step by Step

**Key Point:** `range(2, n)` generates: 2, 3, 4 (NOT 1, 2, 3)

**Trace for n=5:**

| i | fib before | fib[i-1] | fib[i-2] | next_fib | fib after |
|---|-----------|----------|----------|----------|-----------|
| 2 | [0, 1] | fib[1]=1 | fib[0]=0 | 1+0=1 | [0,1,1] |
| 3 | [0, 1, 1] | fib[2]=1 | fib[1]=1 | 1+1=2 | [0,1,1,2] |
| 4 | [0, 1, 1, 2] | fib[3]=2 | fib[2]=1 | 2+1=3 | [0,1,1,2,3] |

**Result:** [0, 1, 1, 2, 3] ✅ Correct!

## Alternative: More Elegant Fibonacci Approach
**Key Differences:**
- Uses negative indexing (`fib[-1]`, `fib[-2]`) instead of explicit indices
- Starts loop from 0, allowing simpler logic
- Uses `str.join()` for cleaner output formatting
- More concise and Pythonic

````python
fib = [0, 1]
n = 5
for i in range(n):
    fib.append(fib[-1] + fib[-2])
# Converting the list of integers to string
print(', '.join(str(e) for e in fib))
# Output: 0, 1, 1, 2, 3, 5, 8
````

## Comparison Analysis

| Aspect | Approach 1 (Index-based) | Approach 2 (Negative Index) |
|--------|--------------------------|---------------------------|
| **Loop Range** | `range(2, n)` - Starts from 2 | `range(n)` - Starts from 0 |
| **Indexing** | Uses `fib[i-1]` and `fib[i-2]` | Uses `fib[-1]` and `fib[-2]` |
| **Readability** | More explicit | More concise/Pythonic |
| **Performance** | Slightly faster (direct indexing) | Nearly identical (negligible difference) |
| **Output Format** | f-string | `str.join()` |

**Why the second approach is better:**
- **Simpler logic**: Doesn't need `range(2, n)` offset calculation
- **More Pythonic**: Uses negative indexing idiomatically
- **No index confusion**: Avoids off-by-one errors with range boundaries
- **Better output**: `', '.join()` is more elegant than list repr

````python
# Q 21. Create a generator to produce first n prime numbers (and prime check function)

def prime_number_checker(n):
    """Check if n is a prime number"""
    if n < 2:
        return False
    
    count = 0
    for i in range(2, n):
        if n % i == 0:
            count += 1
    
    # If count == 0, no divisors found → it's prime!
    if count == 0:
        return True
    else:
        return False

# Test the function
print(f"Is 279 prime? {prime_number_checker(279)}")  # Should be False
print(f"Is 17 prime? {prime_number_checker(17)}")    # Should be True
print(f"Is 2 prime? {prime_number_checker(2)}")      # Should be True

def prime_generator(n):
    """Generate first n prime numbers"""
    prime_list = []  # Use descriptive name instead of 'list'
    i = 2
    
    while len(prime_list) < n:
        if prime_number_checker(i):
            prime_list.append(i)
        i += 1
    
    # Use f-string to format output
    print(f"The first {n} prime numbers are: {prime_list}")
    return prime_list

# Test the generator
prime_generator(10)
````

## When to Use Which?

### Use `*args` (recommended for this use case)
✅ When you want flexible, multiple arguments
```python
sum_all(1, 2, 3)        # Pass individual values
sum_all(10, 20)         # Any number of arguments
print(1, 2, 3)          # Python's print() uses *args
```

### Use `args` (no asterisk)
✅ When caller will pass a collection as one argument
```python
process_list([1, 2, 3])         # Pass a list
iterate_tuple((4, 5, 6))        # Pass a tuple
apply_to_array(np.array([...]))  # Pass an array
```

**Best Practice:** Use `*args` for user-friendly APIs that accept variable arguments!

## Key Clarification: Length vs Number of Arguments

**Important distinction:**

✅ **WITHOUT `*args` (just `args`):**
- You can pass a list of **ANY LENGTH** as input
- But you can only pass **ONE argument** (which is a collection)
- The length is flexible, but the number of arguments is NOT

✅ **WITH `*args`:**
- You can pass **ANY NUMBER** of separate arguments
- Each argument becomes one element in the tuple
- Both the count AND length are flexible

````python
# Example: Comparison of flexibility

# WITHOUT *args (just args)
def sum_without_args(args):
    return sum(args)

# ✅ CAN pass lists of ANY length:
print("Without *args:")
print(sum_without_args([1, 2]))                  # Length 2
print(sum_without_args([1, 2, 3, 4, 5]))        # Length 5
print(sum_without_args([1, 2, 3, 4, 5, 6, 7])) # Length 7

# ❌ CANNOT pass multiple separate arguments:
# sum_without_args(1, 2, 3)  # TypeError!

print("\n" + "="*50 + "\n")

# WITH *args
def sum_with_args(*args):
    return sum(args)

# ✅ CAN pass ANY NUMBER of separate arguments:
print("With *args:")
print(sum_with_args(1, 2))                  # 2 arguments
print(sum_with_args(1, 2, 3, 4, 5))        # 5 arguments
print(sum_with_args(1, 2, 3, 4, 5, 6, 7))  # 7 arguments
````

````python
# Example 1: Mixed types - FAILS at execution
def sum_mixed(args):
    return sum(args)  # Will fail with mixed types

try:
    result = sum_mixed(1, 2, 3, 4, 'abc', {2, 22}, 10)
    print(result)
except TypeError as e:
    print(f"❌ Error: {e}")
    print(f"   *args accepted the arguments, but sum() can't add strings and numbers!")

print("\n" + "="*60 + "\n")

# Example 2: Show what *args receives
def show_args(*args):
    print(f"Received args: {args}")
    print(f"Types: {[type(x).__name__ for x in args]}")
    
show_args(1, 2, 3, 4, 'abc', {2, 22, 2}, 10)
# Notice: *args accepts ALL of them, no errors!
````

````python
# 32. What is python decorator ?

# Decorator lets you add extra behaviour to the function without changing the function's code
# It is a function that takes another function as an input and returns a new function
# https://www.w3schools.com/python/python_decorators.asp

# @changecase is the decorator function
def changecase(func):
  def myinner():
    return func().upper()
  return myinner

@changecase
def myfunction():
  return "Hello Sally"

@changecase
def otherfunction():
  return "I am speed!"

print(myfunction())
print(otherfunction())

# Arguements in the decorated function,

def changecase(func):
  def myinner(x):
    return func(x).upper()
  return myinner

@changecase
def myfunction(nam):
  return "Hello " + nam

print(myfunction("John"))

````

````python
# Function Decorator
def welcome(fx):
    def mfx(*t, **d):
        print("Before hello function")
        fx(*t, **d) # *args to take arguments as tuple, **kwargs to take arguments as dict
        print("Thanks for using the function")
    return mfx

# Decorator function without arguments
@welcome
def hello():
    print("Hello! \n")

# Decorator function with arguments
@welcome
def add(a, b):
    print(a + b)

hello()
add(1, 3)
````

````python
# Class Decorator
class Calculator:
    def __init__(self, func):
        self.function = func
        
    def __call__(self, *t, **d):
        result = self.function(*t, **d)
        return result**2

@Calculator
def add(a, b):
    return a + b

# add = Calculator(add)
print(add(10, 20)) # add.__call__(a,b) since function type is callable
````

````python
# Decorators are used to modify the behavior of functions or methods without
# changing their code. A Decorator function is a function that takes another 
# function as its argument.

def decor_result(result_function):
    def distinction(marks):
        results = [] # List to store results for each element
        for m in marks:
            if m > 75:
                print("DISTINCTION")
            results.append(result_function([m])) # Pass a list with the current mark
        return results # Return the collected results
    return distinction

@decor_result
def result(marks):
    for m in marks:
        if m >= 33:
            pass
        else:
            print("FAIL")
            return "FAIL" # Return FAIL if any element fails
    print("PASS")
    return "PASS" # Return PASS if all elements pass

# Get the results for each element
results = result([45, 78, 80, 34, 66, 90])
print("Results:", results)
````
